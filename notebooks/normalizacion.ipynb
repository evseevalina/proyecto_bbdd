**Decisiones de diseño**

- Se identifica la vertical de cada clase a partir de los proyectos de los csv originales.
- Se supone que cada proyecto pertenece a una sola vertical y que las clases de esa vertical comparten los mismos proyectos.
- No se relaciona directamente Claustro con Clases porque los datos disponibles no permiten identificar todas las clases con certeza.
- No se crea la tabla Aulas porque no se dispone de información.
- Se crean catálogos para mejorar la integridad y la escalabilidad.
- Se asigna un identificador a cada nombre de profesor distinto.

#### 1.5. Modelo relacional definitivo

El modelo final tiene nueve tablas relacionadas mediante claves primarias (PK) y foráneas (FK).

| Tabla | Clave primaria (PK) | Claves foráneas (FK) | Otros atributos |
|---|---|---|---|
| Alumnos | id_alumno | id_clase | Nombre, Email |
| Profesores | id_profesor | — | Nombre |
| Campus | id_campus | — | Campus |
| Verticales | id_vertical | — | Vertical |
| Roles | id_rol | — | Rol |
| Clases | id_clase | id_campus, id_vertical | Promocion, Fecha_comienzo |
| Proyectos | id_proyecto | id_vertical | Proyecto |
| Evaluaciones | id_alumno + id_proyecto | id_alumno, id_proyecto | Evaluacion |
| Claustro | id_claustro | id_profesor, id_campus, id_vertical, id_rol | Promocion, Modalidad |

**Clasificación de las tablas**

| Catálogos | Principales | Asociativas |
|---|---|---|
| Campus | Alumnos | Evaluaciones |
| Verticales | Profesores | Claustro |
| Roles | Clases | |
| | Proyectos | |
| **3 tablas** | **4 tablas** | **2 tablas** |

**Total: 9 tablas.**

**Descripción de las entidades**

| Entidad | Descripción |
|---|---|
| Campus | Ubicaciones de los centros de formacion |
| Verticales | Programas de formación |
| Roles | Funciones del profesorado |
| Clases | Grupos de formación |
| Alumnos | Información de los estudiantes |
| Profesores | Información del profesorado |
| Proyectos | Proyectos académicos de cada vertical |
| Evaluaciones | Resultados de los alumnos en los proyectos |
| Claustro | Asignaciones del profesorado |

**Relaciones principales**

- Cada alumno pertenece a una sola clase y cada clase puede tener varios alumnos.
- Cada clase pertenece a un campus y una vertical y tiene una promoción.
- Cada proyecto pertenece a una vertical.
- Evaluaciones relacionan los alumnos con los proyectos.
- Claustro relaciona a los profesores con los campus, las verticales y los roles. Cada registro también indica la promoción.

**Restricciones de integridad**

- El correo electrónico de cada alumno es único.
- Cada alumno debe tener una clase asignada.
- Cada alumno tiene una única evaluación por proyecto.
- Las claves foráneas garantizan la existencia de los registros relacionados.

**Resultado:** se obtiene un modelo relacional con nueve tablas, menos datos repetidos y una estructura fácil de mantener y ampliar.
